```python
# Kernal Density Estimate plot ( KDE PLOT) 
# how can i visulize audience rating & critics rating . using scatterplot
```

```python
# KDE plot showing the density of Critic Rating vs Audience Rating
k1 = sns.kdeplot(x=movies['CriticRating'], y=movies['AudienceRating'], shade=True, cmap='viridis')

# Plot labels and title
plt.xlabel('Critic Rating')
plt.ylabel('Audience Rating')
plt.title('KDE Plot: Critic vs Audience Rating')
plt.show()
```

```python
# 1 row, 2 columns (so 2 plots side-by-side)
f, ax = plt.subplots(1, 2, figsize=(12, 6))

# Example: Use `ax[0]` for the first subplot and `ax[1]` for the second.

# Plot KDE in the first subplot
sns.kdeplot(x=movies['BudgetMillions'], y=movies['AudienceRating'], fill=True, ax=ax[0])
ax[0].set_title('Budget vs Audience Rating')

# Plot KDE in the second subplot
sns.kdeplot(x=movies['BudgetMillions'], y=movies['CriticRating'], fill=True, ax=ax[1])
ax[1].set_title('Budget vs Critic Rating')

plt.show()
```

```python
# Create subplots
f, axes = plt.subplots(1, 2, figsize=(12, 6))

# KDE plot for Budget vs Audience Rating
k1 = sns.kdeplot(
    x=movies['BudgetMillions'],
    y=movies['AudienceRating'],
    fill=True,
    ax=axes[0]
)
axes[0].set_title('Budget vs Audience Rating')

# KDE plot for Budget vs Critic Rating
k2 = sns.kdeplot(
    x=movies['BudgetMillions'],
    y=movies['CriticRating'],
    fill=True,
    ax=axes[1]
)
axes[1].set_title('Budget vs Critic Rating')

plt.tight_layout()
plt.show()
```

```python
sns.set_style('darkgrid')

# Set up 2x2 subplot grid
f, axes = plt.subplots(2, 2, figsize=(15, 15))

# KDE: Budget vs Audience Rating
sns.kdeplot(
    x=movies['BudgetMillions'],
    y=movies['AudienceRating'],
    fill=True,
    ax=axes[0, 0]
).set(xlim=(-20, 160))

# KDE: Budget vs Critic Rating
sns.kdeplot(
    x=movies['BudgetMillions'],
    y=movies['CriticRating'],
    fill=True,
    ax=axes[0, 1]
).set(xlim=(-20, 160))

# Violin Plot: Critic Ratings by Year for Drama genre
sns.violinplot(
    data=movies[movies['Genre'] == 'Drama'],
    x='Year',
    y='CriticRating',
    ax=axes[1, 0]
)

# Combined KDE: Critic Rating vs Audience Rating
sns.kdeplot(
    x=movies['CriticRating'],
    y=movies['AudienceRating'],
    fill=True,
    cmap='Reds',
    ax=axes[1, 1]
)
sns.kdeplot(
    x=movies['CriticRating'],
    y=movies['AudienceRating'],
    fill=False,
    cmap='Reds',
    ax=axes[1, 1]
)

# Show the complete dashboard
plt.tight_layout()
plt.show()
```
