# EDA:  

**Exploratory Data Analysis (EDA)** is the process of understanding your dataset before diving into modeling or predictions. It helps you:

- Discover patterns

- Spot anomalies

- Test assumptions

- Summarize key characteristics

- Think of EDA as “getting to know your data.”

# 🛠️ Why is EDA Important?
- Ensures your data is clean and usable

- Helps identify outliers and missing values

- Lets you find correlations and distributions

- Guides feature selection for models
